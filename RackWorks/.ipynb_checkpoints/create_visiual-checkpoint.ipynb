```python
import openpyxl
from openpyxl.styles import PatternFill, Border, Side, Alignment, Font
import os

def generate_visual():
    # Delete the file if it already exists
    filename = "rack_layout_top_down.xlsx"
    if os.path.exists(filename):
        os.remove(filename)
        print(f"Existing file '{filename}' deleted.")

    # Create a new workbook
    wb = openpyxl.Workbook()
    ws = wb.active
    ws.title = "Rack Layout"

    # Set column widths
    ws.column_dimensions['A'].width = 25
    ws.column_dimensions['B'].width = 15
    ws.column_dimensions['C'].width = 15
    ws.column_dimensions['D'].width = 15

    # Define styles
    header_font = Font(bold=True)
    border = Border(left=Side(style='thin'), right=Side(style='thin'), 
                    top=Side(style='thin'), bottom=Side(style='thin'))
    center_aligned = Alignment(horizontal='center')

    # Define colors for different node types
    color_map = {
        'compute_nodes': 'FF9999',  # Light red
        'GPU_nodes': '99CCFF',      # Light blue
        'LAN': '99FF99',            # Light green
    }

    # Write headers
    headers = ["Rack Position", "Device Name", "Type", "Height"]
    for col, header in enumerate(headers, start=1):
        cell = ws.cell(row=1, column=col, value=header)
        cell.font = header_font
        cell.border = border
        cell.alignment = center_aligned

    # Create a dictionary to track occupied positions
    occupied_positions = {}
    position_numbers = {}  # To store the actual rack position numbers

    # First pass: Identify all positions and their heights
    for device, posy in node_positions.items():
        if 'LAN_' in device:
            # Handle LAN devices
            lan_parts = device.split('__')
            lan_type = lan_parts[0]
            switch_model = lan_parts[1].split('_')[0]
            height = 2  # Default height for switches

            for switch in LANs[lan_type]['switch']:
                if switch['model'] == switch_model:
                    height = switch['height']
                    break
            pos = posy + height -1 
            actual_pos = 43 - pos  # Convert to top-down numbering
            print('sssss',pos,posy, actual_pos)
            occupied_positions[actual_pos] = {
                'device': device,
                'type': 'LAN Switch',
                'height': height,
                'color': color_map.get('LAN', 'FFFFFF'),
                'original_pos': pos
            }
        else:
            # Handle other nodes
            node_type = device.split('_')[0] + '_' + device.split('_')[1]
            height = colors_info[node_type]['height']
            pos = posy + height -1 
            actual_pos = 43 - pos  # Convert to top-down numbering
            print('ddddd',pos,posy,  actual_pos)
            occupied_positions[actual_pos] = {
                'device': device,
                'type': node_type.replace('_', ' ').title(),
                'height': height,
                'color': color_map.get(node_type, 'FFFFFF'),
                'original_pos': pos
            }

    # Calculate physical position numbers accounting for device heights
    physical_positions = {}
    current_physical_pos = 42
    occupied_label = dict()
    for rack_pos in range(1,43):
        if rack_pos in occupied_positions:
            device_info = occupied_positions[rack_pos]
            height = device_info['height']
            
            # Assign the physical position to the top of the device
            physical_positions[rack_pos] = current_physical_pos
            occupied_label[rack_pos] = str(current_physical_pos-height+1)+'-'+str(current_physical_pos)
            # Mark all occupied positions by this device
            for i in range(1, height):
                physical_positions[rack_pos + i] = None  # Mark as occupied
            
            # Decrement the physical position counter by 1 (not by height)
            current_physical_pos -= height
        else:
            physical_positions[rack_pos] = current_physical_pos
            current_physical_pos -= 1
    #print(physical_positions)
    #print(occupied_positions)
    #sssss
    # Second pass: Write data to cells and handle merging
    current_row = 2
    for rack_pos in range(1, 43):
        # Skip if this is part of a multi-slot device (not the top)
        if rack_pos in occupied_positions and physical_positions[rack_pos] is None:
            current_row += 1
            continue
            
        # Get the physical position number
        if rack_pos in physical_positions:
            position_number = str(physical_positions[rack_pos])
            position_number = occupied_label.get(rack_pos,str(physical_positions[rack_pos]))
            if position_number.startswith('-') or position_number.startswith('0'):
                position_number = None
            
        else:
            position_number = None

        # Write rack position number
        if position_number is not None:
            ws.cell(row=current_row, column=1, value=position_number)
            ws.cell(row=current_row, column=1).border = border
            ws.cell(row=current_row, column=1).alignment = center_aligned

        if rack_pos in occupied_positions and physical_positions[rack_pos] is not None and physical_positions[rack_pos] > 0 :
            device_info = occupied_positions[rack_pos]
            height = device_info['height']
           

            # Write device info
            ws.cell(row=current_row, column=2, value=device_info['device'])
            ws.cell(row=current_row, column=3, value=device_info['type'])
            ws.cell(row=current_row, column=4, value=height)

            # Apply styling to all columns
            fill = PatternFill(start_color=device_info['color'], end_color=device_info['color'], fill_type='solid')
            for col in range(1, 5):
                ws.cell(row=current_row, column=col).fill = fill
                ws.cell(row=current_row, column=col).border = border
                ws.cell(row=current_row, column=col).alignment = center_aligned

            # Merge cells if height > 1
            if height > 1:
                for col in range(1, 5):
                    ws.merge_cells(start_row=current_row, start_column=col,
                                 end_row=current_row + height - 1, end_column=col)
                
                # Skip the merged rows in the main loop
                current_row += height - 1
        else:
            # Empty position
            if  physical_positions[rack_pos] > 0:
                for col in range(2, 5):
                    ws.cell(row=current_row, column=col, value="")
                    ws.cell(row=current_row, column=col).border = border

        current_row += 1

    # Save the workbook
    wb.save(filename)
    print(f"Excel file '{filename}' has been created with positions from 42 at top to 1 at bottom.")
```
